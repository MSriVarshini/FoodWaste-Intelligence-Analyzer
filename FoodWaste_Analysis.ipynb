```python
df["Temperature"]=df["Temperature"].fillna(df["Temperature"].median())
df["Customers"] = df["Customers"].fillna(df["Customers"].median())

df["Cost_Per_Unit"] = df["Cost_Per_Unit"].fillna(df["Cost_Per_Unit"].median())
```

```python
df.drop_duplicates(inplace=True)
```

HERE WE HAVE FINISHED THE DATA CLEANING BY REMOVING THE DULPICATES , FILLING THE NULL VALUES BY MEDIAN VALUES . 


NOW LETS DO FEATURE ENGINEERING , TO KEEP IMP FEATURES AND REMOVE UNWANTED ONES.......

```python
df["Month"]=df["Date"].dt.month
df["Month_Name"] = df["Date"].dt.month_name()
df["Day"] = df["Date"].dt.day
df["Day_Name"] = df["Date"].dt.day_name()
```

```python
df.columns
```

```python
df["Waste_Percentage"] = (
    df["Quantity_Wasted"] / df["Quantity_Produced"]
) * 100
```

```python
df["Financial_Loss"] = (
    df["Quantity_Wasted"] * df["Cost_Per_Unit"]
)
```

```python
df["Total_Revenue"] = (
    df["Quantity_Sold"] * df["Revenue_Per_Unit"]
)
```

```python
df["Potential_Revenue"] = (
    df["Quantity_Produced"] * df["Revenue_Per_Unit"]
)
```

```python
conditions = [
    df["Waste_Percentage"] < 15,
    df["Waste_Percentage"].between(15, 30),
    df["Waste_Percentage"] > 30
]

choices = ["Low", "Medium", "High"]

df["Waste_Risk"] = np.select(
    conditions,
    choices,
    default="Unknown"
)
```

nowww lets move to EDA process with pandasss


```python
#lets find total food produced , food sold, food wasted


print("total food produced :",df["Quantity_Produced"].sum())
print("total food sold:",df["Quantity_Sold"].sum())
print("total food wasted:",df["Quantity_Wasted"].sum())
```

```python
overall_waste_percentage = (
    df["Quantity_Wasted"].sum() /
    df["Quantity_Produced"].sum()
) * 100

print(f"Overall Waste Percentage: {overall_waste_percentage:.2f}%")
```

```python
# which food is wasted more


category_waste = (
    df.groupby("Food_Category")["Quantity_Wasted"]
    .sum()
    .sort_values(ascending=False)
)

category_waste
```

```python
location_waste = (
    df.groupby("Location")["Quantity_Wasted"]
    .sum()
    .sort_values(ascending=False)
)

location_waste
```

```python
top_wasted_items = (
    df.groupby("Food_Item")["Quantity_Wasted"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_wasted_items
```

```python
category_loss = (
    df.groupby("Food_Category")["Financial_Loss"]
    .sum()
    .sort_values(ascending=False)
)

category_loss
```
